```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
import os

# Ensure the /mnt/data directory exists
# In a typical Colab environment, /mnt/data is not writable by default
# so we'll simulate it by creating dummy files in the current working directory
# and then reading them. If you intend to use actual files, ensure they are uploaded
# and accessible at the specified paths.

# Create dummy 'uci_diabetes.csv' file
uci_data = {
    'Glucose': np.random.randint(70, 200, 50),
    'BloodPressure': np.random.randint(60, 120, 50),
    'SkinThickness': np.random.randint(10, 60, 50),
    'Insulin': np.random.randint(0, 800, 50),
    'BMI': np.random.uniform(18, 50, 50),
    'DiabetesPedigreeFunction': np.random.uniform(0.1, 2.0, 50),
    'Age': np.random.randint(20, 80, 50),
    'Outcome': np.random.randint(0, 2, 50)
}
df_uci = pd.DataFrame(uci_data)
df_uci.to_csv('uci_diabetes.csv', index=False)

# Create dummy 'pima_diabetes.csv' file
pima_data = {
    'Glucose': np.random.randint(70, 200, 50),
    'BloodPressure': np.random.randint(60, 120, 50),
    'SkinThickness': np.random.randint(10, 60, 50),
    'Insulin': np.random.randint(0, 800, 50),
    'BMI': np.random.uniform(18, 50, 50),
    'DiabetesPedigreeFunction': np.random.uniform(0.1, 2.0, 50),
    'Age': np.random.randint(20, 80, 50),
    'Outcome': np.random.randint(0, 2, 50)
}
df_pima = pd.DataFrame(pima_data)
df_pima.to_csv('pima_diabetes.csv', index=False)

# Import Datasets
# Reading from the current directory where dummy files are created
uci_diabetes = pd.read_csv('uci_diabetes.csv')
pima_diabetes = pd.read_csv('pima_diabetes.csv')

# Display Dataset Samples
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())
print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())

# Define Relevant Numerical Columns
numerical_columns = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI",
                    "DiabetesPedigreeFunction", "Age"]

# Univariate Analysis Function
def univariate_analysis(df, columns):
    stats = {}
    for col in columns:
        # Ensure the column exists before trying to calculate stats
        if col in df.columns:
            stats[col] = {
                "Mean": np.mean(df[col]),
                "Median": np.median(df[col]),
                "Mode": df[col].mode()[0] if not df[col].empty else np.nan, # Handle empty series for mode
                "Variance": np.var(df[col], ddof=1) if len(df[col]) > 1 else np.nan,
                "Standard Deviation": np.std(df[col], ddof=1) if len(df[col]) > 1 else np.nan,
                "Skewness": skew(df[col]) if len(df[col]) > 2 else np.nan, # Skewness needs at least 3 points
                "Kurtosis": kurtosis(df[col]) if len(df[col]) > 3 else np.nan # Kurtosis needs at least 4 points
            }
        else:
            print(f"Warning: Column '{col}' not found in DataFrame.")
            stats[col] = {
                "Mean": np.nan,
                "Median": np.nan,
                "Mode": np.nan,
                "Variance": np.nan,
                "Standard Deviation": np.nan,
                "Skewness": np.nan,
                "Kurtosis": np.nan
            }
    return pd.DataFrame(stats).T

# Perform Univariate Analysis
uci_stats = univariate_analysis(uci_diabetes, numerical_columns)
pima_stats = univariate_analysis(pima_diabetes, numerical_columns)

# Display Results
print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)
print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)
```
